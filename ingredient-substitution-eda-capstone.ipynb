```python
import pandas as pd
import json
from collections import Counter
import matplotlib.pyplot as plt
from wordcloud import WordCloud
```

```python
# File path
file_path = "/kaggle/input/ingredient-substitution/train.json"

# Load the data from the JSON file
with open(file_path, 'r') as f:
    data = json.load(f)
```

```python
# Convert the JSON data to a DataFrame
df = pd.DataFrame(data)
```

```python
# Basic Info
print("Number of records:", len(df))
print("Unique cuisines:", df['cuisine'].nunique())
print("Cuisines count:\n", df['cuisine'].value_counts())
```

```python
# Count all ingredients
all_ingredients = [ingredient for ingredients in df['ingredients'] for ingredient in ingredients]
ingredient_counts = Counter(all_ingredients)
```

```python
# Most common ingredients
most_common_ingredients = ingredient_counts.most_common(10)
print("Top 10 ingredients:", most_common_ingredients)
```

```python
# Plot Cuisine Distribution
df['cuisine'].value_counts().plot(kind='bar', title='Cuisine Distribution', figsize=(8, 6))
plt.xlabel('Cuisine')
plt.ylabel('Number of Recipes')
plt.show()
```
